```python
import pandas as pd
```

```python
ipl_data = pd.read_csv(r"/home/pavandeep_369/Downloads/IPL.csv")  # change encoding=... if needed
```

```python
staging_data=ipl_data[ipl_data['innings']==2]
```

```python
staging_data_sorted = staging_data.sort_values(["ball_no"], ascending=[False])
Matches_data = staging_data_sorted.groupby("match_id", as_index=False).head(1)
Matches_data =Matches_data.sort_values(["match_id"], ascending=[True])
```

```python
Matches_data.head()
```

```python
Matches_data=Matches_data.drop(columns=['match_type', 'event_name', 'innings',
        'over', 'ball', 'ball_no', 'batter',
       'bat_pos', 'runs_batter', 'balls_faced', 'bowler', 'valid_ball',
       'runs_extras', 'runs_total', 'runs_bowler', 'runs_not_boundary',
       'extra_type', 'non_striker', 'non_striker_pos', 'wicket_kind',
       'player_out', 'fielders', 'runs_target', 'review_batter',
       'team_reviewed', 'review_decision', 'umpire', 'umpires_call',
        'day', 'month', 'season',
       'gender', 'team_type', 'superover_winner', 'result_type', 'method',
       'balls_per_over', 'overs', 'event_match_no', 'stage', 'match_number',
       'team_runs', 'team_balls', 'team_wicket', 'new_batter',
       'power_surge_start', 'batter_runs', 'batter_balls', 'bowler_wicket',
       'batting_partners', 'striker_out', 'next_batter'])
```
